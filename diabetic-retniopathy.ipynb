```python
import os
import shutil
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img

# =========================
# CONFIG
# =========================
src_dir = "/kaggle/input/diabetic-retinopathy"      # original (read-only)
dst_dir = "/kaggle/working/augmented_data"          # writable destination
target_count = 2000                                 # desired number of images per class

# =========================
# Step 1: Copy dataset to working directory
# =========================
if not os.path.exists(dst_dir):
    print("📦 Copying dataset to writable directory...")
    shutil.copytree(src_dir, dst_dir)
else:
    print("✅ Dataset already copied to working directory.")

# =========================
# Step 2: Count original images
# =========================
classes = sorted(os.listdir(dst_dir))
before_counts = {cls: len(os.listdir(os.path.join(dst_dir, cls))) for cls in classes}

print("\n📊 BEFORE AUGMENTATION:")
for k, v in before_counts.items():
    print(f"{k}: {v} images")

# =========================
# Step 3: Define augmentation generator
# =========================
datagen = ImageDataGenerator(
    rotation_range=25,
    width_shift_range=0.15,
    height_shift_range=0.15,
    shear_range=0.1,
    zoom_range=0.2,
    horizontal_flip=True,
    brightness_range=[0.8, 1.2],
    fill_mode='nearest'
)

# =========================
# Step 4: Augment low-sample classes
# =========================
for cls in classes:
    class_path = os.path.join(dst_dir, cls)
    images = os.listdir(class_path)
    n_images = len(images)
    
    if n_images < target_count:
        print(f"\n🧪 Augmenting '{cls}' (current: {n_images}) ...")
        i = 0
        while n_images + i < target_count:
            img_path = os.path.join(class_path, images[i % n_images])
            img = load_img(img_path)
            x = img_to_array(img)
            x = x.reshape((1,) + x.shape)
            
            for batch in datagen.flow(
                x,
                batch_size=1,
                save_to_dir=class_path,
                save_prefix='aug',
                save_format='jpg'
            ):
                i += 1
                if n_images + i >= target_count:
                    break
        print(f"✅ Finished augmenting '{cls}' ({target_count} total images).")

# =========================
# Step 5: Count after augmentation
# =========================
after_counts = {cls: len(os.listdir(os.path.join(dst_dir, cls))) for cls in classes}

print("\n📈 AFTER AUGMENTATION:")
for k, v in after_counts.items():
    print(f"{k}: {v} images")

# =========================
# Step 6: Plot Before & After (Pie + Bar)
# =========================
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Pie Chart - Before
axes[0].pie(
    before_counts.values(),
    labels=before_counts.keys(),
    autopct='%1.1f%%',
    startangle=90,
    colors=plt.cm.Pastel1.colors
)
axes[0].set_title("Before Augmentation")

# Pie Chart - After
axes[1].pie(
    after_counts.values(),
    labels=after_counts.keys(),
    autopct='%1.1f%%',
    startangle=90,
    colors=plt.cm.Set3.colors
)
axes[1].set_title("After Augmentation")

plt.show()

# Bar Chart Comparison
plt.figure(figsize=(10,6))
bar_width = 0.35
x = range(len(classes))

plt.bar(x, before_counts.values(), width=bar_width, label='Before', alpha=0.7)
plt.bar([i + bar_width for i in x], after_counts.values(), width=bar_width, label='After', alpha=0.7)
plt.xticks([i + bar_width / 2 for i in x], classes, rotation=30)
plt.ylabel("Number of Images")
plt.title("Before vs After Augmentation")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Full ViT training + evaluation pipeline (PyTorch + timm)
# Put this in a Kaggle notebook cell and run.
import os
import random
import shutil
from glob import glob
from collections import Counter
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# install timm if missing
import sys
try:
    import timm
except Exception:
    !pip install timm --quiet
    import timm

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from sklearn.model_selection import train_test_split
from sklearn.metrics import (confusion_matrix, classification_report,
                             roc_curve, auc, precision_recall_curve, average_precision_score)
from sklearn.preprocessing import label_binarize
from sklearn.calibration import calibration_curve
import seaborn as sns

# -----------------------------
# Config
# -----------------------------
SRC_DIR = "/kaggle/working/augmented_data"  # your writable dataset copy
CLASSES = sorted(os.listdir(SRC_DIR))  # e.g. ["Healthy", "Mild DR", ...]
NUM_CLASSES = len(CLASSES)
SAMPLES_PER_CLASS = 1000   # take up to 1000 images per class
IMAGE_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 8         # change to 20+ for final training
LR = 2e-4
WEIGHT_DECAY = 1e-4
NUM_WORKERS = 4
RANDOM_SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)

# -----------------------------
# Gather samples (limit per class)
# -----------------------------
random.seed(RANDOM_SEED)
all_image_paths = []
all_labels = []

for idx, cls in enumerate(CLASSES):
    cls_dir = os.path.join(SRC_DIR, cls)
    imgs = sorted(glob(os.path.join(cls_dir, "*.*")))
    # filter common image types
    imgs = [p for p in imgs if p.lower().endswith(('.jpg', '.jpeg', '.png'))]
    if len(imgs) == 0:
        raise RuntimeError(f"No images found for class folder: {cls_dir}")
    selected = imgs if len(imgs) <= SAMPLES_PER_CLASS else random.sample(imgs, SAMPLES_PER_CLASS)
    all_image_paths += selected
    all_labels += [idx] * len(selected)

print("Total collected images:", len(all_image_paths))
print("Per-class counts:", Counter(all_labels))

# -----------------------------
# Split into train / val (80/20 stratified)
# -----------------------------
train_paths, val_paths, train_y, val_y = train_test_split(
    all_image_paths, all_labels, test_size=0.20, random_state=RANDOM_SEED, stratify=all_labels
)
print("Train:", len(train_paths), "Val:", len(val_paths))

# -----------------------------
# Dataset & Transforms
# -----------------------------
train_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomRotation(20),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

val_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

class RetinopathyDataset(Dataset):
    def __init__(self, paths, labels, transform=None):
        self.paths = paths
        self.labels = labels
        self.transform = transform
    def __len__(self):
        return len(self.paths)
    def __getitem__(self, idx):
        p = self.paths[idx]
        img = Image.open(p).convert('RGB')
        if self.transform:
            img = self.transform(img)
        label = self.labels[idx]
        return img, label, p  # return path for later analysis

train_ds = RetinopathyDataset(train_paths, train_y, transform=train_transforms)
val_ds   = RetinopathyDataset(val_paths, val_y,   transform=val_transforms)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# -----------------------------
# Model: ViT (from timm)
# -----------------------------
model = timm.create_model('vit_base_patch16_224', pretrained=True, num_classes=NUM_CLASSES)
model.to(DEVICE)

# Freeze backbone initially (optional). Here we fine-tune full model but can freeze early:
# for param in model.patch_embed.parameters():
#     param.requires_grad = False

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

# -----------------------------
# Training loop
# -----------------------------
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for imgs, labels, _ in loader:
        imgs = imgs.to(device); labels = labels.to(device)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * imgs.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total += imgs.size(0)
    return running_loss / total, correct / total

def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    all_probs = []
    all_preds = []
    all_labels = []
    all_paths = []
    with torch.no_grad():
        for imgs, labels, paths in loader:
            imgs = imgs.to(device)
            outputs = model(imgs)
            probs = torch.softmax(outputs, dim=1)
            preds = outputs.argmax(dim=1).cpu().numpy()
            all_probs.append(probs.cpu().numpy())
            all_preds.extend(preds.tolist())
            all_labels.extend(labels.numpy().tolist())
            all_paths.extend(paths)
            loss = criterion(outputs, labels.to(device)).item()
            running_loss += loss * imgs.size(0)
            total += imgs.size(0)
            correct += (preds == labels.numpy()).sum().item()
    all_probs = np.vstack(all_probs)
    return running_loss/total, correct/total, np.array(all_preds), np.array(all_labels), all_probs, all_paths

best_val_acc = 0.0
history = {'train_loss':[], 'train_acc':[], 'val_loss':[], 'val_acc':[]}

for epoch in range(1, EPOCHS+1):
    train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion, DEVICE)
    val_loss, val_acc, val_preds, val_labels, val_probs, val_paths = evaluate(model, val_loader, criterion, DEVICE)
    scheduler.step()
    history['train_loss'].append(train_loss); history['train_acc'].append(train_acc)
    history['val_loss'].append(val_loss); history['val_acc'].append(val_acc)
    print(f"Epoch {epoch}/{EPOCHS} - train_loss:{train_loss:.4f} acc:{train_acc:.4f} | val_loss:{val_loss:.4f} acc:{val_acc:.4f}")
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), "/kaggle/working/vit_best.pth")
        print("Saved best model.")

# load best for evaluation
model.load_state_dict(torch.load("/kaggle/working/vit_best.pth", map_location=DEVICE))

# -----------------------------
# Final eval on validation
# -----------------------------
val_loss, val_acc, val_preds, val_labels, val_probs, val_paths = evaluate(model, val_loader, criterion, DEVICE)
print("Final VAL acc:", val_acc)

# -----------------------------
# Confusion Matrix + Classification Report
# -----------------------------
cm = confusion_matrix(val_labels, val_preds)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=CLASSES, yticklabels=CLASSES, cmap='Blues')
plt.xlabel("Predicted"); plt.ylabel("True"); plt.title("Confusion Matrix")
plt.show()

print("\nClassification Report:\n")
print(classification_report(val_labels, val_preds, target_names=CLASSES, digits=4))

# -----------------------------
# ROC Curves (One-vs-Rest) + AUC per class
# -----------------------------
# Binarize labels
y_bin = label_binarize(val_labels, classes=list(range(NUM_CLASSES)))  # shape (n_samples, n_classes)
plt.figure(figsize=(10,8))
colors = plt.cm.get_cmap('tab10', NUM_CLASSES)
fpr = dict(); tpr = dict(); roc_auc = dict()
for i in range(NUM_CLASSES):
    fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], val_probs[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
    plt.plot(fpr[i], tpr[i], lw=2, label=f"{CLASSES[i]} (AUC = {roc_auc[i]:.3f})")
plt.plot([0,1],[0,1],'k--', lw=1)
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curves (One-vs-Rest)")
plt.legend(loc='best')
plt.show()

# -----------------------------
# Precision-Recall Curves + AP
# -----------------------------
plt.figure(figsize=(10,8))
pr_auc = {}
for i in range(NUM_CLASSES):
    precision, recall, _ = precision_recall_curve(y_bin[:, i], val_probs[:, i])
    ap = average_precision_score(y_bin[:, i], val_probs[:, i])
    pr_auc[i] = ap
    plt.plot(recall, precision, lw=2, label=f"{CLASSES[i]} (AP = {ap:.3f})")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.title("Precision-Recall Curves")
plt.legend(loc='best')
plt.show()

# -----------------------------
# Calibration plot (reliability diagram)
# -----------------------------
# We compute calibration for the predicted probability of the predicted class for each sample
predicted_probas = val_probs.max(axis=1)
predicted_classes = val_probs.argmax(axis=1)
true_labels_bin = (predicted_classes == val_labels).astype(int)  # correctness
prob_true, prob_pred = calibration_curve(true_labels_bin, predicted_probas, n_bins=10)

plt.figure(figsize=(6,6))
plt.plot(prob_pred, prob_true, marker='o', label='Model')
plt.plot([0,1],[0,1],'k--', label='Perfect calibration')
plt.xlabel("Predicted probability")
plt.ylabel("Fraction of positives")
plt.title("Calibration plot (predicted class probabilities)")
plt.legend()
plt.show()

# -----------------------------
# Misclassification analysis: show top misclassified images
# -----------------------------
# Collect misclassified indices and sort by model confidence (descending)
mis_idx = np.where(val_preds != val_labels)[0]
print("Total misclassified:", len(mis_idx))
# compute confidence of predicted class for each misclassified sample
mis_conf = val_probs[mis_idx, val_preds[mis_idx]]
# pick top N most confident wrong predictions
N = 12
if len(mis_idx) > 0:
    topk = mis_idx[np.argsort(-mis_conf)][:N]
    plt.figure(figsize=(15,9))
    for i, idx in enumerate(topk):
        p = val_paths[idx]
        img = Image.open(p).convert('RGB').resize((224,224))
        plt.subplot(3,4,i+1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(f"T:{CLASSES[val_labels[idx]]}\nP:{CLASSES[val_preds[idx]]}\nConf:{val_probs[idx,val_preds[idx]]:.2f}")
    plt.suptitle("Top confident misclassifications (true / predicted / conf)")
    plt.show()
else:
    print("No misclassifications found.")

# -----------------------------
# Occlusion sensitivity / simple ablation study for selected misclassified images
# -----------------------------
# Sliding window occlusion: for a selected image, replace patch by mean color and measure drop in probability for true class.
def occlusion_map(model, pil_img, transform, device, target_class, occ_size=32, occ_stride=16):
    model.eval()
    img_orig = transform(pil_img).unsqueeze(0).to(device)
    baseline_probs = torch.softmax(model(img_orig), dim=1).cpu().numpy()[0]
    baseline = baseline_probs[target_class]
    w, h = pil_img.size
    heatmap = np.zeros(((h - occ_size)//occ_stride + 1, (w - occ_size)//occ_stride + 1))
    y_positions = range(0, h-occ_size+1, occ_stride)
    x_positions = range(0, w-occ_size+1, occ_stride)
    for i, y in enumerate(y_positions):
        for j, x in enumerate(x_positions):
            img_copy = pil_img.copy()
            # draw occlusion rectangle with mean color
            mean_color = tuple(map(int, np.array(img_copy).mean(axis=(0,1))))
            for yy in range(y, y+occ_size):
                for xx in range(x, x+occ_size):
                    if 0 <= yy < h and 0 <= xx < w:
                        img_copy.putpixel((xx,yy), mean_color)
            inp = transform(img_copy).unsqueeze(0).to(device)
            prob = torch.softmax(model(inp), dim=1).cpu().numpy()[0][target_class]
            heatmap[i,j] = baseline - prob  # drop in prob
    return heatmap, x_positions, y_positions, baseline

# Run occlusion on up to 3 misclassified images (or some random corrects as illustrative)
selected_for_occlusion = []
if len(mis_idx) >= 3:
    selected_for_occlusion = list(mis_idx[:3])
elif len(mis_idx) > 0:
    selected_for_occlusion = list(mis_idx[:min(3,len(mis_idx))])
else:
    # pick some correctly predicted samples as demo
    correct_idx = np.where(val_preds == val_labels)[0][:3]
    selected_for_occlusion = list(correct_idx)

for idx in selected_for_occlusion:
    pil = Image.open(val_paths[idx]).convert('RGB').resize((IMAGE_SIZE, IMAGE_SIZE))
    target_cls = val_labels[idx]
    heatmap, x_pos, y_pos, baseline = occlusion_map(model, pil, val_transforms, DEVICE, target_cls, occ_size=32, occ_stride=16)
    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1); plt.imshow(pil); plt.title(f"True: {CLASSES[target_cls]} Pred: {CLASSES[val_preds[idx]]}\nBaseline prob: {baseline:.3f}"); plt.axis('off')
    plt.subplot(1,2,2); sns.heatmap(heatmap, cmap='viridis'); plt.title("Occlusion map (prob drop)"); plt.xlabel("x patches"); plt.ylabel("y patches")
    plt.show()

# -----------------------------
# Advanced result analysis: per-class metrics, support, and saving results
# -----------------------------
# Per-class AUC and AP
for i in range(NUM_CLASSES):
    print(f"{CLASSES[i]}: AUC={roc_auc[i]:.4f}, AP={pr_auc[i]:.4f}, Val_support={(val_labels==i).sum()}")

# Save evaluation arrays for further analysis
np.save("/kaggle/working/val_probs.npy", val_probs)
np.save("/kaggle/working/val_preds.npy", val_preds)
np.save("/kaggle/working/val_labels.npy", val_labels)
with open("/kaggle/working/class_names.txt", "w") as f:
    f.write("\n".join(CLASSES))

print("Saved evaluation outputs to /kaggle/working/")
```

```python
# Full ViT training + evaluation pipeline (PyTorch + timm)
# Put this in a Kaggle notebook cell and run.
import os
import random
import shutil
from glob import glob
from collections import Counter
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# install timm if missing
import sys
try:
    import timm
except Exception:
    !pip install timm --quiet
    import timm

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from sklearn.model_selection import train_test_split
from sklearn.metrics import (confusion_matrix, classification_report,
                             roc_curve, auc, precision_recall_curve, average_precision_score)
from sklearn.preprocessing import label_binarize
from sklearn.calibration import calibration_curve
import seaborn as sns

# -----------------------------
# Config
# -----------------------------
SRC_DIR = "/kaggle/working/augmented_data"  # your writable dataset copy
CLASSES = sorted(os.listdir(SRC_DIR))  # e.g. ["Healthy", "Mild DR", ...]
NUM_CLASSES = len(CLASSES)
SAMPLES_PER_CLASS = 1000   # take up to 1000 images per class
IMAGE_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 8         # change to 20+ for final training
LR = 2e-4
WEIGHT_DECAY = 1e-4
NUM_WORKERS = 4
RANDOM_SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)

# -----------------------------
# Gather samples (limit per class)
# -----------------------------
random.seed(RANDOM_SEED)
all_image_paths = []
all_labels = []

for idx, cls in enumerate(CLASSES):
    cls_dir = os.path.join(SRC_DIR, cls)
    imgs = sorted(glob(os.path.join(cls_dir, "*.*")))
    # filter common image types
    imgs = [p for p in imgs if p.lower().endswith(('.jpg', '.jpeg', '.png'))]
    if len(imgs) == 0:
        raise RuntimeError(f"No images found for class folder: {cls_dir}")
    selected = imgs if len(imgs) <= SAMPLES_PER_CLASS else random.sample(imgs, SAMPLES_PER_CLASS)
    all_image_paths += selected
    all_labels += [idx] * len(selected)

print("Total collected images:", len(all_image_paths))
print("Per-class counts:", Counter(all_labels))

# -----------------------------
# Split into train / val (80/20 stratified)
# -----------------------------
train_paths, val_paths, train_y, val_y = train_test_split(
    all_image_paths, all_labels, test_size=0.20, random_state=RANDOM_SEED, stratify=all_labels
)
print("Train:", len(train_paths), "Val:", len(val_paths))

# -----------------------------
# Dataset & Transforms
# -----------------------------
train_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomRotation(20),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

val_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

class RetinopathyDataset(Dataset):
    def __init__(self, paths, labels, transform=None):
        self.paths = paths
        self.labels = labels
        self.transform = transform
    def __len__(self):
        return len(self.paths)
    def __getitem__(self, idx):
        p = self.paths[idx]
        img = Image.open(p).convert('RGB')
        if self.transform:
            img = self.transform(img)
        label = self.labels[idx]
        return img, label, p  # return path for later analysis

train_ds = RetinopathyDataset(train_paths, train_y, transform=train_transforms)
val_ds   = RetinopathyDataset(val_paths, val_y,   transform=val_transforms)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# -----------------------------
# Model: ViT (from timm)
# -----------------------------
model = timm.create_model('vit_base_patch16_224', pretrained=True, num_classes=NUM_CLASSES)
model.to(DEVICE)

# Freeze backbone initially (optional). Here we fine-tune full model but can freeze early:
# for param in model.patch_embed.parameters():
#     param.requires_grad = False

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

# -----------------------------
# Training loop
# -----------------------------
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for imgs, labels, _ in loader:
        imgs = imgs.to(device); labels = labels.to(device)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * imgs.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total += imgs.size(0)
    return running_loss / total, correct / total

def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    all_probs = []
    all_preds = []
    all_labels = []
    all_paths = []
    with torch.no_grad():
        for imgs, labels, paths in loader:
            imgs = imgs.to(device)
            labels_device = labels.to(device)
            outputs = model(imgs)
            probs = torch.softmax(outputs, dim=1)
            preds = outputs.argmax(dim=1)
            
            # Convert to numpy AFTER moving to CPU and detaching
            all_probs.append(probs.cpu().numpy())
            all_preds.extend(preds.cpu().numpy().tolist())
            all_labels.extend(labels.numpy().tolist())
            all_paths.extend(paths)
            
            loss = criterion(outputs, labels_device).item()
            running_loss += loss * imgs.size(0)
            total += imgs.size(0)
            correct += (preds.cpu().numpy() == labels.numpy()).sum().item()
            
    all_probs = np.vstack(all_probs)
    return running_loss/total, correct/total, np.array(all_preds), np.array(all_labels), all_probs, all_paths

best_val_acc = 0.0
history = {'train_loss':[], 'train_acc':[], 'val_loss':[], 'val_acc':[]}

for epoch in range(1, EPOCHS+1):
    train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion, DEVICE)
    val_loss, val_acc, val_preds, val_labels, val_probs, val_paths = evaluate(model, val_loader, criterion, DEVICE)
    scheduler.step()
    history['train_loss'].append(train_loss); history['train_acc'].append(train_acc)
    history['val_loss'].append(val_loss); history['val_acc'].append(val_acc)
    print(f"Epoch {epoch}/{EPOCHS} - train_loss:{train_loss:.4f} acc:{train_acc:.4f} | val_loss:{val_loss:.4f} acc:{val_acc:.4f}")
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), "/kaggle/working/vit_best.pth")
        print("Saved best model.")

# load best for evaluation
model.load_state_dict(torch.load("/kaggle/working/vit_best.pth", map_location=DEVICE))

# -----------------------------
# Final eval on validation
# -----------------------------
val_loss, val_acc, val_preds, val_labels, val_probs, val_paths = evaluate(model, val_loader, criterion, DEVICE)
print("Final VAL acc:", val_acc)

# -----------------------------
# Confusion Matrix + Classification Report
# -----------------------------
cm = confusion_matrix(val_labels, val_preds)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=CLASSES, yticklabels=CLASSES, cmap='Blues')
plt.xlabel("Predicted"); plt.ylabel("True"); plt.title("Confusion Matrix")
plt.show()

print("\nClassification Report:\n")
print(classification_report(val_labels, val_preds, target_names=CLASSES, digits=4))

# -----------------------------
# ROC Curves (One-vs-Rest) + AUC per class
# -----------------------------
# Binarize labels
y_bin = label_binarize(val_labels, classes=list(range(NUM_CLASSES)))  # shape (n_samples, n_classes)
plt.figure(figsize=(10,8))
colors = plt.cm.get_cmap('tab10', NUM_CLASSES)
fpr = dict(); tpr = dict(); roc_auc = dict()
for i in range(NUM_CLASSES):
    fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], val_probs[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
    plt.plot(fpr[i], tpr[i], lw=2, label=f"{CLASSES[i]} (AUC = {roc_auc[i]:.3f})")
plt.plot([0,1],[0,1],'k--', lw=1)
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curves (One-vs-Rest)")
plt.legend(loc='best')
plt.show()

# -----------------------------
# Precision-Recall Curves + AP
# -----------------------------
plt.figure(figsize=(10,8))
pr_auc = {}
for i in range(NUM_CLASSES):
    precision, recall, _ = precision_recall_curve(y_bin[:, i], val_probs[:, i])
    ap = average_precision_score(y_bin[:, i], val_probs[:, i])
    pr_auc[i] = ap
    plt.plot(recall, precision, lw=2, label=f"{CLASSES[i]} (AP = {ap:.3f})")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.title("Precision-Recall Curves")
plt.legend(loc='best')
plt.show()

# -----------------------------
# Calibration plot (reliability diagram)
# -----------------------------
# We compute calibration for the predicted probability of the predicted class for each sample
predicted_probas = val_probs.max(axis=1)
predicted_classes = val_probs.argmax(axis=1)
true_labels_bin = (predicted_classes == val_labels).astype(int)  # correctness
prob_true, prob_pred = calibration_curve(true_labels_bin, predicted_probas, n_bins=10)

plt.figure(figsize=(6,6))
plt.plot(prob_pred, prob_true, marker='o', label='Model')
plt.plot([0,1],[0,1],'k--', label='Perfect calibration')
plt.xlabel("Predicted probability")
plt.ylabel("Fraction of positives")
plt.title("Calibration plot (predicted class probabilities)")
plt.legend()
plt.show()

# -----------------------------
# Misclassification analysis: show top misclassified images
# -----------------------------
# Collect misclassified indices and sort by model confidence (descending)
mis_idx = np.where(val_preds != val_labels)[0]
print("Total misclassified:", len(mis_idx))
# compute confidence of predicted class for each misclassified sample
mis_conf = val_probs[mis_idx, val_preds[mis_idx]]
# pick top N most confident wrong predictions
N = 12
if len(mis_idx) > 0:
    topk = mis_idx[np.argsort(-mis_conf)][:N]
    plt.figure(figsize=(15,9))
    for i, idx in enumerate(topk):
        p = val_paths[idx]
        img = Image.open(p).convert('RGB').resize((224,224))
        plt.subplot(3,4,i+1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(f"T:{CLASSES[val_labels[idx]]}\nP:{CLASSES[val_preds[idx]]}\nConf:{val_probs[idx,val_preds[idx]]:.2f}")
    plt.suptitle("Top confident misclassifications (true / predicted / conf)")
    plt.show()
else:
    print("No misclassifications found.")

# -----------------------------
# Occlusion sensitivity / simple ablation study for selected misclassified images
# -----------------------------
# Sliding window occlusion: for a selected image, replace patch by mean color and measure drop in probability for true class.
def occlusion_map(model, pil_img, transform, device, target_class, occ_size=32, occ_stride=16):
    model.eval()
    img_orig = transform(pil_img).unsqueeze(0).to(device)
    with torch.no_grad():
        baseline_probs = torch.softmax(model(img_orig), dim=1).cpu().numpy()[0]
    baseline = baseline_probs[target_class]
    w, h = pil_img.size
    heatmap = np.zeros(((h - occ_size)//occ_stride + 1, (w - occ_size)//occ_stride + 1))
    y_positions = range(0, h-occ_size+1, occ_stride)
    x_positions = range(0, w-occ_size+1, occ_stride)
    for i, y in enumerate(y_positions):
        for j, x in enumerate(x_positions):
            img_copy = pil_img.copy()
            # draw occlusion rectangle with mean color
            mean_color = tuple(map(int, np.array(img_copy).mean(axis=(0,1))))
            for yy in range(y, y+occ_size):
                for xx in range(x, x+occ_size):
                    if 0 <= yy < h and 0 <= xx < w:
                        img_copy.putpixel((xx,yy), mean_color)
            inp = transform(img_copy).unsqueeze(0).to(device)
            with torch.no_grad():
                prob = torch.softmax(model(inp), dim=1).cpu().numpy()[0][target_class]
            heatmap[i,j] = baseline - prob  # drop in prob
    return heatmap, x_positions, y_positions, baseline

# Run occlusion on up to 3 misclassified images (or some random corrects as illustrative)
selected_for_occlusion = []
if len(mis_idx) >= 3:
    selected_for_occlusion = list(mis_idx[:3])
elif len(mis_idx) > 0:
    selected_for_occlusion = list(mis_idx[:min(3,len(mis_idx))])
else:
    # pick some correctly predicted samples as demo
    correct_idx = np.where(val_preds == val_labels)[0][:3]
    selected_for_occlusion = list(correct_idx)

for idx in selected_for_occlusion:
    pil = Image.open(val_paths[idx]).convert('RGB').resize((IMAGE_SIZE, IMAGE_SIZE))
    target_cls = val_labels[idx]
    heatmap, x_pos, y_pos, baseline = occlusion_map(model, pil, val_transforms, DEVICE, target_cls, occ_size=32, occ_stride=16)
    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1); plt.imshow(pil); plt.title(f"True: {CLASSES[target_cls]} Pred: {CLASSES[val_preds[idx]]}\nBaseline prob: {baseline:.3f}"); plt.axis('off')
    plt.subplot(1,2,2); sns.heatmap(heatmap, cmap='viridis'); plt.title("Occlusion map (prob drop)"); plt.xlabel("x patches"); plt.ylabel("y patches")
    plt.show()

# -----------------------------
# Advanced result analysis: per-class metrics, support, and saving results
# -----------------------------
# Per-class AUC and AP
for i in range(NUM_CLASSES):
    print(f"{CLASSES[i]}: AUC={roc_auc[i]:.4f}, AP={pr_auc[i]:.4f}, Val_support={(val_labels==i).sum()}")

# Save evaluation arrays for further analysis
np.save("/kaggle/working/val_probs.npy", val_probs)
np.save("/kaggle/working/val_preds.npy", val_preds)
np.save("/kaggle/working/val_labels.npy", val_labels)
with open("/kaggle/working/class_names.txt", "w") as f:
    f.write("\n".join(CLASSES))

print("Saved evaluation outputs to /kaggle/working/")
```

```python
# Full ViT training + evaluation pipeline (PyTorch + timm)
# Put this in a Kaggle notebook cell and run.
import os
import random
import shutil
from glob import glob
from collections import Counter
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# install timm if missing
import sys
try:
    import timm
except Exception:
    !pip install timm --quiet
    import timm

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from sklearn.model_selection import train_test_split
from sklearn.metrics import (confusion_matrix, classification_report,
                             roc_curve, auc, precision_recall_curve, average_precision_score)
from sklearn.preprocessing import label_binarize
from sklearn.calibration import calibration_curve
import seaborn as sns

# -----------------------------
# Config
# -----------------------------
SRC_DIR = "/kaggle/working/augmented_data"  # your writable dataset copy
CLASSES = sorted(os.listdir(SRC_DIR))  # e.g. ["Healthy", "Mild DR", ...]
NUM_CLASSES = len(CLASSES)
SAMPLES_PER_CLASS = 2000   # take up to 2000 images per class
IMAGE_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 8         # change to 20+ for final training
LR = 2e-4
WEIGHT_DECAY = 1e-4
NUM_WORKERS = 4
RANDOM_SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)

# -----------------------------
# Gather samples (limit per class)
# -----------------------------
random.seed(RANDOM_SEED)
all_image_paths = []
all_labels = []

for idx, cls in enumerate(CLASSES):
    cls_dir = os.path.join(SRC_DIR, cls)
    imgs = sorted(glob(os.path.join(cls_dir, "*.*")))
    # filter common image types
    imgs = [p for p in imgs if p.lower().endswith(('.jpg', '.jpeg', '.png'))]
    if len(imgs) == 0:
        raise RuntimeError(f"No images found for class folder: {cls_dir}")
    selected = imgs if len(imgs) <= SAMPLES_PER_CLASS else random.sample(imgs, SAMPLES_PER_CLASS)
    all_image_paths += selected
    all_labels += [idx] * len(selected)

print("Total collected images:", len(all_image_paths))
print("Per-class counts:", Counter(all_labels))

# -----------------------------
# Split into train / val (80/20 stratified)
# -----------------------------
train_paths, val_paths, train_y, val_y = train_test_split(
    all_image_paths, all_labels, test_size=0.20, random_state=RANDOM_SEED, stratify=all_labels
)
print("Train:", len(train_paths), "Val:", len(val_paths))

# -----------------------------
# Dataset & Transforms
# -----------------------------
train_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomRotation(20),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

val_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

class RetinopathyDataset(Dataset):
    def __init__(self, paths, labels, transform=None):
        self.paths = paths
        self.labels = labels
        self.transform = transform
    def __len__(self):
        return len(self.paths)
    def __getitem__(self, idx):
        p = self.paths[idx]
        img = Image.open(p).convert('RGB')
        if self.transform:
            img = self.transform(img)
        label = self.labels[idx]
        return img, label, p  # return path for later analysis

train_ds = RetinopathyDataset(train_paths, train_y, transform=train_transforms)
val_ds   = RetinopathyDataset(val_paths, val_y,   transform=val_transforms)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# -----------------------------
# Model: ViT (from timm)
# -----------------------------
model = timm.create_model('vit_base_patch16_224', pretrained=True, num_classes=NUM_CLASSES)
model.to(DEVICE)

# Freeze backbone initially (optional). Here we fine-tune full model but can freeze early:
# for param in model.patch_embed.parameters():
#     param.requires_grad = False

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

# -----------------------------
# Training loop
# -----------------------------
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for imgs, labels, _ in loader:
        imgs = imgs.to(device); labels = labels.to(device)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * imgs.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total += imgs.size(0)
    return running_loss / total, correct / total

def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    all_probs = []
    all_preds = []
    all_labels = []
    all_paths = []
    with torch.no_grad():
        for imgs, labels, paths in loader:
            imgs = imgs.to(device)
            labels_device = labels.to(device)
            outputs = model(imgs)
            probs = torch.softmax(outputs, dim=1)
            preds = outputs.argmax(dim=1)
            
            # Convert to numpy AFTER moving to CPU and detaching
            all_probs.append(probs.cpu().numpy())
            all_preds.extend(preds.cpu().numpy().tolist())
            all_labels.extend(labels.numpy().tolist())
            all_paths.extend(paths)
            
            loss = criterion(outputs, labels_device).item()
            running_loss += loss * imgs.size(0)
            total += imgs.size(0)
            correct += (preds.cpu().numpy() == labels.numpy()).sum().item()
            
    all_probs = np.vstack(all_probs)
    return running_loss/total, correct/total, np.array(all_preds), np.array(all_labels), all_probs, all_paths

best_val_acc = 0.0
history = {'train_loss':[], 'train_acc':[], 'val_loss':[], 'val_acc':[]}

for epoch in range(1, EPOCHS+1):
    train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion, DEVICE)
    val_loss, val_acc, val_preds, val_labels, val_probs, val_paths = evaluate(model, val_loader, criterion, DEVICE)
    scheduler.step()
    history['train_loss'].append(train_loss); history['train_acc'].append(train_acc)
    history['val_loss'].append(val_loss); history['val_acc'].append(val_acc)
    print(f"Epoch {epoch}/{EPOCHS} - train_loss:{train_loss:.4f} acc:{train_acc:.4f} | val_loss:{val_loss:.4f} acc:{val_acc:.4f}")
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), "/kaggle/working/vit_best.pth")
        print("Saved best model.")

# load best for evaluation
model.load_state_dict(torch.load("/kaggle/working/vit_best.pth", map_location=DEVICE))

# -----------------------------
# Final eval on validation
# -----------------------------
val_loss, val_acc, val_preds, val_labels, val_probs, val_paths = evaluate(model, val_loader, criterion, DEVICE)
print("Final VAL acc:", val_acc)

# -----------------------------
# Confusion Matrix + Classification Report
# -----------------------------
cm = confusion_matrix(val_labels, val_preds)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=CLASSES, yticklabels=CLASSES, cmap='Blues')
plt.xlabel("Predicted"); plt.ylabel("True"); plt.title("Confusion Matrix")
plt.show()

print("\nClassification Report:\n")
print(classification_report(val_labels, val_preds, target_names=CLASSES, digits=4))

# -----------------------------
# ROC Curves (One-vs-Rest) + AUC per class
# -----------------------------
# Binarize labels
y_bin = label_binarize(val_labels, classes=list(range(NUM_CLASSES)))  # shape (n_samples, n_classes)
plt.figure(figsize=(10,8))
colors = plt.cm.get_cmap('tab10', NUM_CLASSES)
fpr = dict(); tpr = dict(); roc_auc = dict()
for i in range(NUM_CLASSES):
    fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], val_probs[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
    plt.plot(fpr[i], tpr[i], lw=2, label=f"{CLASSES[i]} (AUC = {roc_auc[i]:.3f})")
plt.plot([0,1],[0,1],'k--', lw=1)
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curves (One-vs-Rest)")
plt.legend(loc='best')
plt.show()

# -----------------------------
# Precision-Recall Curves + AP
# -----------------------------
plt.figure(figsize=(10,8))
pr_auc = {}
for i in range(NUM_CLASSES):
    precision, recall, _ = precision_recall_curve(y_bin[:, i], val_probs[:, i])
    ap = average_precision_score(y_bin[:, i], val_probs[:, i])
    pr_auc[i] = ap
    plt.plot(recall, precision, lw=2, label=f"{CLASSES[i]} (AP = {ap:.3f})")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.title("Precision-Recall Curves")
plt.legend(loc='best')
plt.show()

# -----------------------------
# Calibration plot (reliability diagram)
# -----------------------------
# We compute calibration for the predicted probability of the predicted class for each sample
predicted_probas = val_probs.max(axis=1)
predicted_classes = val_probs.argmax(axis=1)
true_labels_bin = (predicted_classes == val_labels).astype(int)  # correctness
prob_true, prob_pred = calibration_curve(true_labels_bin, predicted_probas, n_bins=10)

plt.figure(figsize=(6,6))
plt.plot(prob_pred, prob_true, marker='o', label='Model')
plt.plot([0,1],[0,1],'k--', label='Perfect calibration')
plt.xlabel("Predicted probability")
plt.ylabel("Fraction of positives")
plt.title("Calibration plot (predicted class probabilities)")
plt.legend()
plt.show()

# -----------------------------
# Misclassification analysis: show top misclassified images
# -----------------------------
# Collect misclassified indices and sort by model confidence (descending)
mis_idx = np.where(val_preds != val_labels)[0]
print("Total misclassified:", len(mis_idx))
# compute confidence of predicted class for each misclassified sample
mis_conf = val_probs[mis_idx, val_preds[mis_idx]]
# pick top N most confident wrong predictions
N = 12
if len(mis_idx) > 0:
    topk = mis_idx[np.argsort(-mis_conf)][:N]
    plt.figure(figsize=(15,9))
    for i, idx in enumerate(topk):
        p = val_paths[idx]
        img = Image.open(p).convert('RGB').resize((224,224))
        plt.subplot(3,4,i+1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(f"T:{CLASSES[val_labels[idx]]}\nP:{CLASSES[val_preds[idx]]}\nConf:{val_probs[idx,val_preds[idx]]:.2f}")
    plt.suptitle("Top confident misclassifications (true / predicted / conf)")
    plt.show()
else:
    print("No misclassifications found.")

# -----------------------------
# Occlusion sensitivity / simple ablation study for selected misclassified images
# -----------------------------
# Sliding window occlusion: for a selected image, replace patch by mean color and measure drop in probability for true class.
def occlusion_map(model, pil_img, transform, device, target_class, occ_size=32, occ_stride=16):
    model.eval()
    img_orig = transform(pil_img).unsqueeze(0).to(device)
    with torch.no_grad():
        baseline_probs = torch.softmax(model(img_orig), dim=1).cpu().numpy()[0]
    baseline = baseline_probs[target_class]
    w, h = pil_img.size
    heatmap = np.zeros(((h - occ_size)//occ_stride + 1, (w - occ_size)//occ_stride + 1))
    y_positions = range(0, h-occ_size+1, occ_stride)
    x_positions = range(0, w-occ_size+1, occ_stride)
    for i, y in enumerate(y_positions):
        for j, x in enumerate(x_positions):
            img_copy = pil_img.copy()
            # draw occlusion rectangle with mean color
            mean_color = tuple(map(int, np.array(img_copy).mean(axis=(0,1))))
            for yy in range(y, y+occ_size):
                for xx in range(x, x+occ_size):
                    if 0 <= yy < h and 0 <= xx < w:
                        img_copy.putpixel((xx,yy), mean_color)
            inp = transform(img_copy).unsqueeze(0).to(device)
            with torch.no_grad():
                prob = torch.softmax(model(inp), dim=1).cpu().numpy()[0][target_class]
            heatmap[i,j] = baseline - prob  # drop in prob
    return heatmap, x_positions, y_positions, baseline

# Run occlusion on up to 3 misclassified images (or some random corrects as illustrative)
selected_for_occlusion = []
if len(mis_idx) >= 3:
    selected_for_occlusion = list(mis_idx[:3])
elif len(mis_idx) > 0:
    selected_for_occlusion = list(mis_idx[:min(3,len(mis_idx))])
else:
    # pick some correctly predicted samples as demo
    correct_idx = np.where(val_preds == val_labels)[0][:3]
    selected_for_occlusion = list(correct_idx)

for idx in selected_for_occlusion:
    pil = Image.open(val_paths[idx]).convert('RGB').resize((IMAGE_SIZE, IMAGE_SIZE))
    target_cls = val_labels[idx]
    heatmap, x_pos, y_pos, baseline = occlusion_map(model, pil, val_transforms, DEVICE, target_cls, occ_size=32, occ_stride=16)
    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1); plt.imshow(pil); plt.title(f"True: {CLASSES[target_cls]} Pred: {CLASSES[val_preds[idx]]}\nBaseline prob: {baseline:.3f}"); plt.axis('off')
    plt.subplot(1,2,2); sns.heatmap(heatmap, cmap='viridis'); plt.title("Occlusion map (prob drop)"); plt.xlabel("x patches"); plt.ylabel("y patches")
    plt.show()

# -----------------------------
# Advanced result analysis: per-class metrics, support, and saving results
# -----------------------------
# Per-class AUC and AP
for i in range(NUM_CLASSES):
    print(f"{CLASSES[i]}: AUC={roc_auc[i]:.4f}, AP={pr_auc[i]:.4f}, Val_support={(val_labels==i).sum()}")

# Save evaluation arrays for further analysis
np.save("/kaggle/working/val_probs.npy", val_probs)
np.save("/kaggle/working/val_preds.npy", val_preds)
np.save("/kaggle/working/val_labels.npy", val_labels)
with open("/kaggle/working/class_names.txt", "w") as f:
    f.write("\n".join(CLASSES))

print("Saved evaluation outputs to /kaggle/working/")
```
